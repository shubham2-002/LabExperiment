```python
person ={
    "first_name":"Satyam",
    "last_name":"Sharma",
    "age":26,
    "sex":"male",
    "married":False
}

print(person["first_name"])
print(person["sex"])
```

```python
for key in person.keys():
    print(key)
```

```python
for value in person.values():
    print(value)
```

```python
for key , value in person.items():
    print(f"{key} :- {value}")
```

```python
# copy copy dictonary
individual  = person.copy()
print(individual)

print(person is individual)
print(person == individual)
```

```python
individual.clear()
print(individual)

new_user = {}.fromkeys(['name','score','email'],'unkonwn')
print(new_user)
```

**Data Modeling**

```python
playlist={
    'title':"Patangonis Bus",
    "authour":"HUman",
    "songs":[
        {
        'song_title':"Song_one",
        'artis':['Artist1','Artist2'],
        'duration':4.21
        },
        {
           'song_title':"Song_two",
        'artis':['Artist1'],
        'duration':2.53 
        },
         {
        'song_title':"Song_three",
        'artis':['Artist1','Artist2','Artist3'],
        'duration':3.43 
        }
    ]
}
```

```python
total_duration=0.0
for song in playlist['songs']:
    total_duration += song['duration']
    print(song['song_title'])
print(total_duration)
```

**Dict Comprehension**

```python
nums={
    'a':1,
    'b':2,
    'c':3,
}
```

```python
sq_nums = {key:value **2 for key,value in nums.items()}
print(sq_nums)
```

```python
sq_numbers = {num:num**2 for num in [1,2,3,4,5]}
print(sq_numbers)
```

```python
str1 ='ABC'
str2 ='123'
combo = {str1[i]:str2[i] for i in  range(0,len(str1))}
print(combo)
```

```python
num_list = [1,2,3,4,5,6]
parity = {num:('even' if num %2==0 else 'odd') for num in num_list}
print(parity)
```
