```python
import cmath
```

```python

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

if a == 0:
    print("Not a quadratic equation.")
else:
    D = b**2 - 4*a*c
    
D = pow(b,2)-(4*a*c)

x1 =(-b + cmath.sqrt(D))/(2*a)
x2 =(-b - cmath.sqrt(D))/(2*a)

print("Discriminant =", D)
print("Root 1 =", x1)
print("Root 2 =", x2)
```

```python
print("1. Binary to Decimal")
print("2. Decimal to Binary")
print("3. Decimal to Hexadecimal")
print("4. Hexadecimal to Decimal")
print("5. Binary to Hexadecimal")
print("6. Hexadecimal to Binary")

choose = int(input("Choose the option"))
if choose==1:
    n =int(input("Enter the binary No"))
    print("Decimal",int(n,2))
    # bin_to_deci()
elif choose==2:
    n =int(input("Enter the Decimal No"))
    print("Binary:", bin(n))
    # bin_to_deci()
elif choose==3:
    n =int(input("Enter the Decimal No"))
    print("Hexadecimal:", hex(n))
    # bin_to_deci()
elif choose==4:
    n =int(input("Enter the hexa No"))
    print("Decimal",int(n,16))
    # bin_to_deci()
elif choose==5:
    n =int(input("Enter the Binary No"))
    deci = int(n, 2)
    print("Hexadecimal: ", hex(deci))
    # bin_to_deci()
elif choose==6:
    n =int(input("Enter the Hexa No"))
    deci = int(n,16)
    print("Binary:",bin(deci,2))
    # bin_to_deci()
else:
    print("Invalid CHoice")
```
