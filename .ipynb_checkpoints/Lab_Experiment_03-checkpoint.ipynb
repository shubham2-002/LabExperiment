```python
import numpy as np
```

```python
def matrixCreation(r,c,Mat):
    for i in range(r):
        row=[]
        for j in range(c):
            x = int(input(f"Enter element {i},{j}:"))
            row.append(x)
        Mat.append(row)
    for row in Mat:
        print(row)
```

```python
print("======For matrix A======")
row1 = int(input("Enter No of Row "))
col1 = int(input("Enter No of col"))
MatA = []
matrixCreation(row1, col1, MatA)

print("\n======For matrix B=======")
row2 = int(input("Enter No of Row "))
col2 = int(input("Enter No of col"))
MatB = []
matrixCreation(row2, col2, MatB)

if col1 != row2:
    print("Multpliication not possible")
else:
    C = [[0] * col2 for i in range(row1)]

    for i in range(row1):
        for j in range(col2):
            for k in range(col1):
                C[i][j] += MatA[i][k] * MatB[k][j]

    print("Product matrix:")
    for row in C:
        print(row)

    if row1 == col2:
        if C == [list(row) for row in zip(*C)]:
            print("Symtric")
        else:
            print("Not symetrc")

        if row1 == 2:
            detC = C[0][0] * C[1][1] - C[0][1] * C[1][0]

            if detC == 0:
                print("Singular matrix")
            else:
                print("Non-singular matrix")
        else:
            print("Singular check implemented only for 2x2 result")
    else:
        print("Result is not square")
```

```python
A = np.array([
    [1, 1, 1],
    [2, -1, 1],
    [1, 2, -1]
], dtype=float)

B = np.array([6, 3, 2], dtype=float)

det_A= np.linalg.det(A)

invA = np.linalg.inv(A)
print(np.round(invA))

X = invA@B
print("\nSolution [x, y, z]:")
print(np.round(X))

# Crammer Rule
A = np.array([
    [1, 1, 1],
    [2, -1, 1],
    [1, 2, -1]
], dtype=float)

B = np.array([6, 3, 2], dtype=float)

D = np.linalg.det(A)

solutions = []

for i in range(3):
    Ai = np.copy(A)
    Ai[:,i]=B
    Di = np.linalg.det(Ai)
    solutions.append(Di/D)
    
print(np.round(solutions))
```

```python

#Calculate determinant and
#inverse of a 3×3 matrix;
A = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])

detA = np.linalg.det(A)
print(round(detA))

invA = np.linalg.inv(A)
print(invA)

ans = A@invA
print(np.round(ans,10))

I=np.eye(3)
print("\nIs A @ invA approximately I?")
print(np.allclose(ans, I))
```
