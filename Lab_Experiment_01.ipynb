```python

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

if a == 0:
    print("Not a quadratic equation.")
else:
    D = b**2 - 4*a*c
    
D = pow(b,2)-(4*a*c)

x1 =(-b + cmath.sqrt(D))/(2*a)
x2 =(-b - cmath.sqrt(D))/(2*a)

print("Discriminant =", D)
print("Root 1 =", x1)
print("Root 2 =", x2)
```

```python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1)+fib(n-2)

n=int(input("Fib sequn upto:"))
while(n>=0):
    print(fib(n))
    n=n-1
```

```python
def fact(n):
    if n==0 or n==1:
        return 1;
    else:
        return n*fact(n-1)
        
n=int(input("Fib sequn upto:"))
while(n>=0):
    print(fact(n))
    n=n-1
```
