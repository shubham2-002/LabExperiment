```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3])
y = np.array([1,2,3])
print("Addition:", x + y)

print("Subtraction:", x - y)


print("Scalar multiplication:", 2 * x)


print("Dot product:", np.dot(x, y))

print("Cross product:", np.cross(x, y))

print("Magnitude of X:", np.linalg.norm(x))
print("Magnitude of Y:", np.linalg.norm(y))
```

```python

A = np.array([3, 4])
B = np.array([1, 0])

cos_t = np.dot(A,B)/(np.linalg.norm(A)*np.linalg.norm(B))
angle = np.degrees(np.arccos(np.clip(cos_t,-1,1)))

projeAB = np.dot(A,B)/np.linalg.norm(B)

print("Angle btwn vetcor A and B",round(angle,4))
print("Projection of A onto B",projeAB)
```

```python
A1 = np.array([1, 4])
B1 = np.array([5, 4])


proj2 = (np.dot(A1, B1) / np.dot(B1, B1)) * B1

print("2D projection:", np.round(proj2, 2))
plt.figure(figsize=(7, 6))
plt.quiver(0, 0, A1[0], A1[1],
           angles="xy", scale_units="xy", scale=1,
           color="blue", label="Vector A")
plt.quiver(0, 0, B1[0], B1[1],
           angles="xy", scale_units="xy", scale=1,
           color="red", label="Vector B")
plt.quiver(0, 0, proj2[0], proj2[1],
           angles="xy", scale_units="xy", scale=1,
           color="black", label="Projection of A onto B")

plt.xlim(-1, 6)
plt.ylim(-1, 6)
plt.grid(True)
plt.legend()
plt.title("2D Vector Projection")
plt.show()
```

```python
A3 = np.array([3, 4, 2])
B3 = np.array([5, 0, 2])

proj3 = np.dot(A3,B3)/(np.dot(B3,B3))*B3
print("3D projection:", np.round(proj3, 2))
fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(111, projection="3d")

ax.quiver(0,0,0,A3[0],A3[1],A3[2] ,color="blue", label="Vector A")
ax.quiver(0,0,0,B3[0],B3[1],B3[2] ,color="Red", label="Vector B")
ax.quiver(0,0,0,proj3[0],proj3[1],proj3[2],color="black", label="Projection of A onto B")

ax.set_xlim(0, 6)
ax.set_ylim(0, 6)
ax.set_zlim(0, 5)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("3D Vector Projection")
```
